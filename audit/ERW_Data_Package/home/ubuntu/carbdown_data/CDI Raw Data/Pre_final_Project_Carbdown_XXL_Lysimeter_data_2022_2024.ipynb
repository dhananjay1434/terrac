```python
import seaborn as sns
import matplotlib.pyplot as plt
df.columns = df.iloc[0]  # Set the first row as headers
df = df[1:]  # Remove the first row (original headers)
df=df.copy()

# Cleanup

df.replace(',', '', regex=True, inplace=True)
df = df.apply(pd.to_numeric, errors='ignore')
df = df.convert_dtypes()

df['Treatment'] = df['Treatment'].astype('category')
df['Date'] = pd.to_datetime(df['Date'],format='mixed',dayfirst=True)
```

```python
# Plot graph of alkalinity measurements over time

# Calculate surface corrected Alkalinity
# divide by 0,48 cm² surface area, convert to mmol
df['HCO3-_umol'] = pd.to_numeric(df['HCO3-_umol'], errors='coerce')
df['surfacecorrected HCO3- mmol'] = df['HCO3-_umol']/ 0.48/1000

plt.figure(figsize=(7, 4))
filtered_df = df[df['Treatment'].isin(['000', '100','200','400',"FINE"])]

# Rename the "Treatment" values before plotting
treatment_mapping = {'000': 'Control', '100': '100 t/ha', '200': '200 t/ha', '400': '400 t/ha', 'FINE': '200 t/ha fine'}
filtered_df['Treatment'] = filtered_df['Treatment'].map(treatment_mapping)

# Create the line plot with renamed treatment values
ax = sns.lineplot(x="Date", y="surfacecorrected HCO3- mmol",
            hue="Treatment",marker='o',
            data=df, errorbar=('ci', 95)
            )
plt.title("XXL Lysimeter Experiment")
plt.xlabel("")
plt.ylabel("Alkalinity (mmol/m²)")
plt.xticks(rotation=25)

# Remove legend title
ax.legend(title=None)
plt.tight_layout()
```

```python
# Plot graph of accumulated alkalinity measurements over time

# Calculate average HCO3- mmol for each month within each treatment group
df['Avg HCO3- mmol'] = df.groupby(['Treatment', 'Date'])['HCO3-_umol'].transform('mean')/1000

# Calculate cumulative sum of the average HCO3- mmol
df['Cumulative Avg HCO3- mmol'] = df.groupby('Treatment')['Avg HCO3- mmol'].cumsum()/4 / 0.48 # divide by 4 replicas and divide by 0,48 cm² surface area

plt.figure(figsize=(7, 4))
filtered_df = df[df['Treatment'].isin(['000', '100','200','400',"FINE"])]

# Rename the "Treatment" values before plotting
treatment_mapping = {'000': 'Control', '100': '100 t/ha', '200': '200 t/ha', '400': '400 t/ha', 'FINE': '200 t/ha fine'}
filtered_df['Treatment'] = filtered_df['Treatment'].map(treatment_mapping)

# Create the line plot with renamed treatment values
ax = sns.lineplot(x="Date", y="Cumulative Avg HCO3- mmol",
            hue="Treatment",marker='o',
            data=filtered_df, errorbar=('ci', 95),
            hue_order=list(treatment_mapping.values())  # Control the legend order
            )

plt.title("XXL Lysimeter Experiment")
plt.xlabel("")
plt.ylabel("Accumulated alkalinity (mmol/m²)")
plt.xticks(rotation=25)

# Remove legend title
ax.legend(title=None)
plt.tight_layout()
```

```python
# Plot graph of alkalinity and accumulated measurements over time, showing all replicates as dots


def plotoneset(df,mask,new_labels,soil):

    df_filtered = df.copy()

    # Calculate the difference in days
    reference_date = pd.to_datetime('2022-05-15')
    df_filtered['days'] = (df_filtered['Date'] - reference_date).dt.days

    df_filtered['HCO3-_umol'] = df_filtered['HCO3-_umol']/0.48  # divide by 0,48 cm² surface area

    # Group by 'Pot-ID'
    grouped = df_filtered.groupby('Pot-ID')

    # Calculate cumulative sum of 'HCO3- mmol' for each group
    df_filtered['Cumulative HCO3- mmol'] = grouped['HCO3-_umol'].cumsum()/1000 # umol to mmol

    plt.figure(figsize=(7, 5))
    ax=sns.lineplot(x="days", y="HCO3-_umol",palette='dark',
            hue="Treatment",
            data=df_filtered, errorbar=('ci', 95)
            )
    # Add data points
    sns.scatterplot(x="days", y="HCO3-_umol", size=0.5,alpha=0.5,
                    hue="Treatment", palette='dark',
                    data=df_filtered, ax=ax, legend=False)
    plt.title("XXL Lysimeter Experiment "+soil)
    plt.xlabel("Days since start of experiment")
    plt.ylabel("Alkalinity mmol/m²")

    handles, labels = ax.get_legend_handles_labels()
    ax.legend(handles=handles, labels=new_labels, title='',fontsize=10)
    plt.tight_layout()
    plt.show()
    plt.figure(figsize=(7, 5))

    ax=sns.lineplot(x="days", y="Cumulative HCO3- mmol",palette='dark',
            hue="Treatment",
            data=df_filtered, errorbar=('ci', 95)
            )
    sns.scatterplot(x="days", y="Cumulative HCO3- mmol", size=0.5,alpha=0.5,
                    hue="Treatment", palette='dark',
                    data=df_filtered, ax=ax, legend=False)
    plt.title("XXL Lysimeter Experiment "+soil)
    plt.xlabel("Days since start of experiment")
    plt.ylabel("Cummulated Alkalinity mmol/m²")
    handles, labels = ax.get_legend_handles_labels()
    ax.legend(handles=handles, labels=new_labels, title='',fontsize=10)
    plt.tight_layout()
    plt.show()
new_labels = ['Control', '100 t/ha', '200 t/ha', '400 t/ha', '200 t/ha FINE']


plotoneset(df,"mask",new_labels,"Fürth 2022 soil")
```
